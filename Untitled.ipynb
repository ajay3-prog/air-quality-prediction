```python
# Step 1: Import Libraries
import os
import json
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Step 2: Load Dataset
csv_path = 'AirQualityUCI.csv'
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        if filename.endswith('.csv'):
            csv_path = os.path.join(dirname, filename)

df = pd.read_csv(csv_path, sep=';', decimal=',')
print("Raw Dataset Shape:", df.shape)
print(df.head())

# Step 3: Data Preprocessing & Dynamic Gas Selection
# Remove empty unnamed columns and empty rows
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
df = df.dropna(how='all').copy()

# Combine Date and Time into Datetime and sort chronologically (hourly sequence)
df['Datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'], format='%d/%m/%Y %H.%M.%S')
df = df.sort_values('Datetime').reset_index(drop=True)

# Automatically detect all gas columns ending with '(GT)'
all_gases = [col for col in df.columns if col.endswith('(GT)')]
valid_gases = []

for gas in all_gases:
    df[gas] = pd.to_numeric(df[gas], errors='coerce')
    missing_pct = ((df[gas] == -200).sum() / len(df)) * 100
    
    # Keep gas if missing values are less than 80%
    if missing_pct < 80:
        valid_gases.append(gas)
        # Preserve original missing flag BEFORE any imputation
        df[gas + '_missing'] = (df[gas] == -200).astype(int)
    else:
        df = df.drop(columns=[gas])

print("Selected Valid Gases:", valid_gases)

# Replace -200 with NaN across numeric columns
sensor_cols = ['PT08.S1(CO)', 'PT08.S2(NMHC)', 'PT08.S3(NOx)', 'PT08.S4(NO2)', 'PT08.S5(O3)']
weather_cols = ['T', 'RH', 'AH']
num_cols = sensor_cols + weather_cols + valid_gases

for col in num_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    df[col] = df[col].replace(-200, np.nan)

# Step 4: Exploratory Data Analysis (Boxplot & Correlation Heatmap)
# 1. Sensor Boxplot
plt.figure(figsize=(8, 4))
df[sensor_cols].dropna().boxplot()
plt.title('Sensor Readings Boxplot (Outlier Detection)')
plt.xticks(rotation=15)
plt.show()

# 2. Correlation Heatmap
corr_matrix = df[sensor_cols + weather_cols + valid_gases].corr()
plt.figure(figsize=(9, 7))
plt.imshow(corr_matrix, cmap='coolwarm', interpolation='nearest')
plt.colorbar(label='Correlation Coefficient')
plt.xticks(range(len(corr_matrix.columns)), corr_matrix.columns, rotation=45, ha='right')
plt.yticks(range(len(corr_matrix.columns)), corr_matrix.columns)
plt.title('Correlation Heatmap (Sensors, Weather & Gases)')
plt.tight_layout()
plt.show()

# Step 5: Feature Engineering (Computed on Continuous Hourly Sequence First)
def create_features(data, target_gas):
    temp = data.copy()
    
    # Time-based features
    temp['Hour'] = temp['Datetime'].dt.hour
    temp['DayOfWeek'] = temp['Datetime'].dt.dayofweek
    temp['Month'] = temp['Datetime'].dt.month
    
    # Forward-fill sensor, weather, and historical series up to 3 hours for lag calculation
    temp_filled = temp[num_cols].ffill(limit=3)
    temp[sensor_cols + weather_cols] = temp_filled[sensor_cols + weather_cols]
    
    # Historical Lagged features computed on continuous hourly rows (Lag_24 = exact 24 hours ago)
    target_series = temp_filled[target_gas]
    temp['Lag_1'] = target_series.shift(1)
    temp['Lag_2'] = target_series.shift(2)
    temp['Lag_3'] = target_series.shift(3)
    temp['Lag_24'] = target_series.shift(24)
    temp['Rolling_Mean_3'] = target_series.shift(1).rolling(window=3).mean()
    
    # Keep target value with limited ffill only for test continuity, but track genuine vs missing using _missing flag
    temp[target_gas] = target_series
    
    time_cols = ['Hour', 'DayOfWeek', 'Month']
    lag_cols = ['Lag_1', 'Lag_2', 'Lag_3', 'Lag_24', 'Rolling_Mean_3']
    feature_cols = sensor_cols + weather_cols + time_cols + lag_cols
    
    # Drop rows where input features or target are still NaN
    temp = temp.dropna(subset=feature_cols + [target_gas]).reset_index(drop=True)
    
    return temp, feature_cols

# Step 6: Random Forest Training, Validation Threshold Calibration & Pickle Saving
os.makedirs('saved_models', exist_ok=True)
registry = {}

for gas in valid_gases:
    clean_name = gas.replace('(GT)', '').strip()
    print(f"\n================ {gas} ================")
    
    temp_df, feat_cols = create_features(df, gas)
    
    # Chronological Split: 70% Train, 10% Validation (for threshold calibration), 20% Future Test
    n = len(temp_df)
    train_end = int(n * 0.70)
    val_end = int(n * 0.80)
    
    train_df = temp_df.iloc[:train_end].copy()
    val_df = temp_df.iloc[train_end:val_end].copy()
    test_df = temp_df.iloc[val_end:].copy()
    
    # Problem 1 Fix: Train ONLY on genuine target observations (exclude imputed target values)
    genuine_train = train_df[train_df[gas + '_missing'] == 0]
    genuine_val = val_df[val_df[gas + '_missing'] == 0]
    
    X_train, y_train = genuine_train[feat_cols], genuine_train[gas]
    X_val, y_val = genuine_val[feat_cols], genuine_val[gas]
    X_test, y_test = test_df[feat_cols], test_df[gas]
    
    # Train Random Forest Regressor on genuine historical data
    rf_model = RandomForestRegressor(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
    rf_model.fit(X_train, y_train)
    
    # Problem 3 Fix: Calibrate Anomaly Threshold using Validation Set (not Test Set)
    val_pred = rf_model.predict(X_val)
    val_rmse = np.sqrt(mean_squared_error(y_val, val_pred))
    threshold = 2.0 * val_rmse
    high_pollution_level = y_train.quantile(0.75)
    
    # Evaluate on Future Test Set (Genuine observations metrics)
    y_pred = rf_model.predict(X_test)
    genuine_test_mask = (test_df[gas + '_missing'] == 0).values
    
    mae = mean_absolute_error(y_test[genuine_test_mask], y_pred[genuine_test_mask])
    mse = mean_squared_error(y_test[genuine_test_mask], y_pred[genuine_test_mask])
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test[genuine_test_mask], y_pred[genuine_test_mask])
    
    print(f"Future Test Metrics -> MAE: {mae:.4f} | MSE: {mse:.4f} | RMSE: {rmse:.4f} | R2: {r2:.4f}")
    print(f"Validation Calibrated Threshold: {threshold:.4f}")
    
    # Pollution Episode Detection & Root-Cause Analysis on Future Period
    test_df['Actual'] = y_test.values
    test_df['Predicted'] = y_pred
    test_df['Residual_Error'] = test_df['Actual'] - test_df['Predicted']
    test_df['Abs_Error'] = test_df['Residual_Error'].abs()
    test_df['Is_Episode'] = test_df['Abs_Error'] > threshold
    
    reasons = []
    for idx, row in test_df.iterrows():
        if row[gas + '_missing'] == 1 and row['Is_Episode']:
            reasons.append('Data Limitation (Missing Target Imputed)')
        elif not row['Is_Episode']:
            reasons.append('Normal Observation')
        elif row['Residual_Error'] > 0 and row['Actual'] >= high_pollution_level:
            reasons.append('Potential Pollution Spike')
        else:
            reasons.append('Unusual Prediction Error')
            
    test_df['Episode_Reason'] = reasons
    episodes = test_df[test_df['Is_Episode']]
    print(f"Total Flagged Events: {len(episodes)}")
    print(episodes['Episode_Reason'].value_counts())
    
    # Save Flagged Episodes CSV
    episodes[['Datetime', 'Actual', 'Predicted', 'Residual_Error', 'Episode_Reason']].to_csv(
        f"saved_models/{clean_name}_episodes.csv", index=False
    )
    
    # Save Model using Pickle
    model_data = {
        'gas_name': gas,
        'model': rf_model,
        'feature_columns': feat_cols,
        'threshold': float(threshold),
        'high_pollution_level': float(high_pollution_level),
        'metrics': {'mae': round(float(mae), 4), 'rmse': round(float(rmse), 4), 'r2': round(float(r2), 4)}
    }
    
    with open(f"saved_models/{clean_name}_model.pkl", 'wb') as f:
        pickle.dump(model_data, f)
        
    registry[clean_name] = {
        'gas_column': gas,
        'model_file': f"{clean_name}_model.pkl",
        'episodes_file': f"{clean_name}_episodes.csv",
        'feature_columns': feat_cols,
        'threshold': round(float(threshold), 4),
        'high_pollution_level': round(float(high_pollution_level), 4),
        'metrics': model_data['metrics']
    }
    
    # Plots for Primary Gas CO(GT): Actual vs Predicted, Feature Importance, and Residual Analysis
    if gas == 'CO(GT)':
        # 1. Actual vs Predicted Graph
        plt.figure(figsize=(10, 4))
        plt.plot(test_df['Actual'].values[:200], label='Actual CO(GT)', color='blue')
        plt.plot(test_df['Predicted'].values[:200], label='Predicted CO(GT)', color='orange', linestyle='--')
        plt.title('Actual vs Predicted CO(GT) on Future Test Period (First 200 Hours)')
        plt.xlabel('Future Test Hours')
        plt.ylabel('CO Concentration')
        plt.legend()
        plt.show()
        
        # 2. Feature Importance Graph
        importances = pd.Series(rf_model.feature_importances_, index=feat_cols).sort_values()
        plt.figure(figsize=(8, 5))
        importances.plot(kind='barh', color='teal')
        plt.title('Random Forest Feature Importance for CO(GT)')
        plt.xlabel('Importance Score')
        plt.show()
        
        # 3. Residual Analysis Graph
        plt.figure(figsize=(7, 4))
        plt.hist(test_df['Residual_Error'], bins=40, color='purple', edgecolor='black', alpha=0.7)
        plt.axvline(threshold, color='red', linestyle='--', label='+Threshold')
        plt.axvline(-threshold, color='red', linestyle='--', label='-Threshold')
        plt.title('Residual Error Distribution (Actual - Predicted)')
        plt.xlabel('Residual Error')
        plt.ylabel('Frequency')
        plt.legend()
        plt.show()

# Save registry file
with open('saved_models/model_registry.json', 'w') as f:
    json.dump(registry, f, indent=4)

print("\nSaved Pickle Files:", os.listdir('saved_models'))
```
